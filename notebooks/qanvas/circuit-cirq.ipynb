# Qanvas circuit

Made with [Qanvas](https://ethan-rivas.github.io/qanvas/), in Cirq. Run each cell with Shift + Enter.

```python
import cirq

q = cirq.LineQubit.range(1)
circuit = cirq.Circuit(
    cirq.Moment(cirq.H(q[0])),
    cirq.Moment(cirq.X(q[0])),
)
```

Run the circuit 1000 times and count each outcome. Outcomes list the measured qubits with q0 first, like Qanvas.

```python
measured = [0]
# The circuit has no measurements, so every qubit is measured.
circuit.append(cirq.measure(q[i], key=f"q{i}") for i in measured)
result = cirq.Simulator().run(circuit, repetitions=1000)
histogram = result.multi_measurement_histogram(keys=[f"q{i}" for i in measured])
counts = {"".join(map(str, bits)): n for bits, n in histogram.items()}
counts
```

```python
import matplotlib.pyplot as plt

outcomes = sorted(counts)
plt.bar(outcomes, [counts[key] for key in outcomes])
plt.xlabel("Outcome (q0 first)")
plt.ylabel("Count")
plt.show()
```
